```python
import pandas as pd




roll_no = "1024160063"

digits = [int(x) for x in roll_no[-2:]]

categories = ["billing", "account", "general"]

personalized_entries = []

for d in digits:
    category = categories[d % 3]

    if d % 3 == 0:
        personalized_entries.append({
            "question": "how do I update my registered mobile number",
            "answer": "You can update your registered mobile number from account settings.",
            "keywords": ["mobile", "update number"],
            "category": category
        })

    elif d % 3 == 1:
        personalized_entries.append({
            "question": "how can I reset my password",
            "answer": "Go to Settings and select Reset Password.",
            "keywords": ["password", "reset login"],
            "category": category
        })

    else:
        personalized_entries.append({
            "question": "how can I check my account details",
            "answer": "You can view your account details from profile section.",
            "keywords": ["account", "details"],
            "category": category
        })


fixed_entries = [

{
"question":"what is the annual fee",
"answer":"The annual fee is Rs 500.",
"keywords":["fee","cost price charge"],
"category":"billing"
},

{
"question":"how to reset password",
"answer":"Go to Settings > Reset Password.",
"keywords":["password","password reset login"],
"category":"account"
},

{
"question":"what are your working hours",
"answer":"We are open 9 AM to 5 PM.",
"keywords":["hours","timing open time"],
"category":"general"
},

{
"question":"how can i pay the fee",
"answer":"You can pay via UPI, card, or net banking.",
"keywords":["pay payment upi fee"],
"category":"billing"
}

]




faq_entries = fixed_entries + personalized_entries


df = pd.DataFrame(faq_entries)

print("Final Knowledge Base")
print(df)




def score_query(query):

    query = query.lower()

    result = []

    for index,row in df.iterrows():

        score = 0

        for word in row["keywords"]:
            if word.lower() in query:
                score += 1

        result.append({
            "question":row["question"],
            "answer":row["answer"],
            "category":row["category"],
            "score":score
        })


    result_df = pd.DataFrame(result)

    result_df = result_df.sort_values(
        by="score",
        ascending=False
    )

    return result_df[result_df["score"]>0]



query = input("\nEnter your query: ")

print("\nMatching Results")
print(score_query(query))




def get_category(category_name):

    return df[df["category"] == category_name]



print("\nCategory Search")
print(get_category("billing"))





choice = input("\nAdd new keyword? y/n : ")

if choice=="y":

    q = input("Enter question: ")

    keyword = input("Enter keyword: ")


    index = df[df["question"]==q].index


    if len(index)>0:

        i=index[0]

        df.at[i,"keywords"].append(keyword)

        print("Keyword added")


    else:
        print("Question not found")



df.to_csv(
    f"{roll_no}_faq_data.csv",
    index=False
)





print("\nFAQ count by category")

print(
    df.groupby("category").size()
)





def improved_score(query):

    query=query.lower()

    answers=[]


    for index,row in df.iterrows():

        score=0


        for key in row["keywords"]:

            if key.lower() in query:

                score+=1


        answers.append(
            {
                "question":row["question"],
                "answer":row["answer"],
                "category":row["category"],
                "score":score
            }
        )


    result=pd.DataFrame(answers)



    max_score=result["score"].max()

    return result[result["score"]==max_score]



print("\nImproved Matching")

print(
    improved_score("fee")
)
```
