```python
import numpy as np
```

```python
arr = np.array([2,4,6,8,10])
add_arr = arr + 2
product_arr = arr*3
divide_arr = arr/2
print(add_arr)
print(product_arr)
print(divide_arr)
```

```python
arr = np.array([1, 2, 3, 6, 4, 5])
print(arr[::-1])
```

```python
x = np.array([1,2,3,4,5,1,2,1,1,1])
values , count = np.unique(x, return_counts=True)
max = np.max(count)
most_frequent = values[count == max]
print("Most Frequent value :",most_frequent)
print("Most Frequent indices:", np.where(x == most_frequent)[0])
```

```python
y = np.array([1,1,1,2,3,4,2,4,3,3])
```

```python
values , count = np.unique(y, return_counts=True)
max = np.max(count)
most_frequent = values[count == max]
print("Most Frequent value :",most_frequent)
print("Most Frequent indices:", np.where(np.isin(y, most_frequent))[0])
```
