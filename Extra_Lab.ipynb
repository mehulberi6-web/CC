```python
import pandas as pd
import numpy as np

# Read CSV file
df = pd.read_csv("student_academic_performance_dataset_100.csv")

# Q2: Locate rows 0, 4, 7 and 8
print(df.loc[[0, 4, 7, 8]])

# Q3: DataFrame navigation
print(df.loc[3:7])           # Rows 3 to 7
print(df.iloc[4:9, 2:5])     # Rows 4 to 8, columns 2 to 4
print(df.iloc[:, 1:4])       # All rows, columns 1 to 3

# Q4: Display first five rows
print(df.head())

# Q5: Delete row 4 and column 3
new_df = df.drop(index=4)
new_df = new_df.drop(columns=df.columns[3])
print(new_df.head())

# Q6(a): Shape
print("Shape:", df.shape)

# Q6(b): Information
df.info()

# Q6(c): Descriptive statistics
print(df.describe())

# Q6(d): First 5 and last 3 rows
print(df.head())
print(df.tail(3))

# Q6(e): Statistics
print("Average Total Marks:", df["Total_Marks"].mean())
print("Total Assignment Marks:", df["Assignment_Marks"].sum())
print("Youngest Age:", df["Age"].min())
print("Highest Attendance:", df["Attendance_Percentage"].max())

# Q6(f): Sort by Total Marks
print(df.sort_values(by="Total_Marks", ascending=False))

# Q6(g): Performance Category
df["Performance_Category"] = np.where(df["Total_Marks"] >= 90, "Excellent",
                             np.where(df["Total_Marks"] >= 75, "Good", "Average"))

# Q6(h): Missing values
print(df.isnull().sum())

# Q6(i): Rename Student_ID
df.rename(columns={"Student_ID": "ID"}, inplace=True)

# Q6(j): Filtering
print(df[df["Age"] > 20])
print(df[df["Department"] == "CSE"])

# Q6(k): Add Bonus Marks column
df["Bonus_Marks"] = df["Total_Marks"] * 0.10

# Q6(l): Save modified dataset
df.to_csv("modified_student_dataset.csv", index=False)

print("Assignment completed successfully!")
```
